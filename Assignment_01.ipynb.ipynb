```python
t = (1, 2, 2, 3, 1, 4, 3)

new = ()

for i in t:
    if i not in new:
        new = new + (i,)

print("Original Tuple:", t)
print("Unique Elements:", new)
```

```python
students = (("Riya", 85), ("Aman", 92),
            ("Soham", 78), ("Priya", 89))

sorted_students = sorted(students,
                         key=lambda x: x[1],
                         reverse=True)

for name, marks in sorted_students:
    print(name, marks)
```

```python
t = ()

for i in range(5):
    n = int(input("Enter an integer: "))
    t = t + (n,)

total = sum(t)
average = total / 5

print("Tuple:", t)
print("Sum:", total)
print("Average:", average)
```

```python
t = (10, 20, 10, 30, 10, 40, 20)

count = t.count(10)

print("Number of occurrences of 10:", count)
```

```python
t = (25, 10, 45, 5, 30, 60)

largest = t[0]
smallest = t[0]

for i in t:
    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

print("Largest Element:", largest)
print("Smallest Element:", smallest)
```

part 2 : Set (4 Programs)

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference:", A - B)
```

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

A = set(list1)
B = set(list2)

common = A & B

print("Common Elements:", common)
```

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

```python
python_students = {"Aman", "Riya", "Soham", "Priya"}

java_students = {"Riya", "Priya", "Rahul", "Ankit"}

result = python_students ^ java_students

print("Students enrolled in exactly one course:")
print(result)
```

part-3 Dictionary (3 programs)

```python
students = {
    "Riya": 85,
    "Aman": 92,
    "Soham": 78,
    "Priya": 89,
    "Rahul": 80
}

for name, marks in students.items():
    print(name, ":", marks)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
d1 = {"Pen": 10, "Pencil": 5, "Book": 3}

d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}

result = d1.copy()

for key, value in d2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

print("Merged Dictionary:", result)
```
